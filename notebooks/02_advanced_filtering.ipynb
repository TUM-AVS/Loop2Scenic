```python
import sys
sys.path.append('..')

from src.pipeline import RAGPipeline
from src.utils import setup_logging

setup_logging(level="INFO")
pipeline = RAGPipeline()
```

## 1. Create Diverse Documents

```python
# Create documents with various tags
documents = [
    {
        "text": "Python Flask is a micro web framework for building web applications.",
        "tags": ["python", "web", "backend"],
        "metadata": {"year": 2024, "category": "tutorial"}
    },
    {
        "text": "React is a JavaScript library for building user interfaces.",
        "tags": ["javascript", "web", "frontend"],
        "metadata": {"year": 2024, "category": "tutorial"}
    },
    {
        "text": "TensorFlow is a machine learning framework developed by Google.",
        "tags": ["python", "ml", "ai"],
        "metadata": {"year": 2024, "category": "reference"}
    }
]

# Ingest documents
texts = [d["text"] for d in documents]
tags = [d["tags"] for d in documents]
metadatas = [d["metadata"] for d in documents]

chunks = pipeline.processor.process_texts(texts=texts, tags=tags, metadatas=metadatas)
doc_ids = pipeline.vectorstore.add_documents(chunks)

print(f"Ingested {len(doc_ids)} documents")
```

## 2. Filter by Tag

```python
# Filter by Python tag
results = pipeline.retriever.retrieve(
    query="programming frameworks",
    filter_tags=["python"],
    return_scores=True
)

print("Documents with 'python' tag:")
for doc, score in results:
    print(f"- {doc.page_content[:60]}...")
    print(f"  Tags: {doc.metadata.get('tags', [])}\n")
```
